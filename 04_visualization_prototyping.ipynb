### Modeling Hypothesis for Streamlit


```python
import pandas as pd, matplotlib.pyplot as plt
m = pd.read_csv("data/processed/merged_dataset_final.csv")
m = m[m["model"] == "embedding_lr"]
fig, ax = plt.subplots(figsize=(3.3, 2.6))
ax.scatter(m["dei_score"], m["volatility"], s=14)
for _, r in m.iterrows():
    ax.annotate(r["company"], (r["dei_score"], r["volatility"]), fontsize=6, xytext=(3, 2), textcoords="offset points")
ax.set_xlabel("DEI score (embeddings + LR)"); ax.set_ylabel("Revenue volatility")
fig.tight_layout(); fig.savefig("reports/figures/score_vs_volatility.pdf")
```
